```python
import pandas as pd
import numpy as np
df = pd.read_csv("messy_dataset.csv")
df.head()
df.info()
df.shape
df.columns
df.describe(include = "all")        
df.dtypes
df.isna().sum()
```

```python
#--------------------- Missing Values Detection --------------------------------------------

df.isna()                                   #identify missing values
df.isna().sum()                             #count missing values
(df.isnull().sum()/ len(df))*100            #percentage of null values
 
```

```python
#--------------------- Duplicate Detection ------------------------------------------------
df.duplicated()
df.duplicated().sum()
df[df.duplicated()]
```

```python

#check every column, if it has less than 20 different values,show me the count of each value.
for col in df.columns:
    if df[col].nunique() < 20:
        print(df[col].value_counts())
        print("-"*20)
```

```python
#Employee_ID
df["Employee_ID"]
df["Employee_ID"].duplicated().sum()
df["Employee_ID"].isna().sum()
df["Employee_ID"].dtype
```

```python
#Employee_Name
df["Employee_Name"]
df["Employee_Name"].isnull().sum()
df["Employee_Name"].duplicated().sum()
df["Employee_Name"].dtype
df["Employee_Name"] = df["Employee_Name"].str.strip().str.title()
df["Employee_Name"]
```

```python
#Department
df["Department"]
df["Department"].isna().sum()
df["Department"].duplicated().sum()
df["Department"] = df["Department"].str.strip().str.upper()
df["Department"].unique()
df["Department"]
```

```python
#Gender
df["Gender"]
df["Gender"].isna().sum()
df["Gender"].unique()
df["Gender"] = df["Gender"].str.strip().str.title()
df["Gender"]
```

```python
#Age
df["Age"]
df["Age"].isna().sum()
df["Age"].dtype
df["Age"] = pd.to_numeric(df["Age"],errors = "coerce")
df.loc[~df["Age"].between(18,70), "Age"] = np.nan
df["Age"] = df["Age"].fillna(df["Age"].median()).astype(int)
df["Age"].isna().sum()
df["Age"].max()
df["Age"]
```

```python
#Salary
df.loc[df["Salary"] < 0, "Salary"] = None
df["Salary"] = df["Salary"].fillna(df["Salary"].median()).astype(int)
df["Salary"]
```

```python
#Joining_Date
df["Joining_Date"]
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"],format="mixed",dayfirst=True,errors = "coerce")
df["Joining_Date"].isna().sum()
df["Joining_Date"]
```

```python
#City
df["City"]
df["City"].unique()
df["City"].isna().sum()
df["City"] = df["City"].str.strip().str.title()
```

```python
#Performance_Rating
df["Performance_Rating"].isna().sum()
df["Performance_Rating"] = pd.to_numeric(df["Performance_Rating"],errors = "coerce")
df.loc[~df["Performance_Rating"].between(1, 5),"Performance_Rating"] = np.nan
df["Performance_Rating"] = df["Performance_Rating"].fillna(df["Performance_Rating"].median()).astype(int)
df["Performance_Rating"]
```

```python
#Experience_Years
df["Experience_Years"]
df["Experience_Years"].isna().sum()
df["Experience_Years"].value_counts()
df.loc[df["Experience_Years"] < 0 ,"Experience_Years"] = np.nan
df["Experience_Years"] = df["Experience_Years"].fillna(df["Experience_Years"].median()).astype(int)
df["Experience_Years"]
 
```

```python
#------------------------------------   FINAL DATA QUALITY ANALYSIS  ------------------------------------------

# 1) Dataset Dimensions:
print("Total rows : ",df.shape[0])
print("Total columns : ",df.shape[1])

# 2) Missing Values:
print("Total Missing Values:", df.isna().sum().sum())

# 3) Duplicate Records:
print("Exact Duplicate Rows:", df.duplicated().sum())
print("Duplicate Employee IDs:", df["Employee_ID"].duplicated().sum())

# 4) Data Types:
print(df.dtypes)

# 5) Data Validation:
print("Invalid Ages : ",((df["Age"] < 18) | df["Age"] > 70).sum())
print("Invalid Salaries : ",((df["Salary"]) < 0).sum())
print("Invalid Performance Ratings:",(~df["Performance_Rating"].between(1, 5)).sum())
print("Invalid Experience:",(df["Experience_Years"] < 0).sum())

# 6) Category consistency:
print("Departments:", df["Department"].unique())
print("Genders:", df["Gender"].unique())
print("Cities:", df["City"].unique())

allowed_cities = ["Karachi","Lahore","Islamabad"]                   #check for allowed values
not_allowed_cities = ~df["City"].isin(allowed_cities)
df[not_allowed_cities]


# 7. Final dataset summary:
print(df.describe(include="all"))



```
